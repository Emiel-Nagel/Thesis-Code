```python
run_dir = "work"
```

```python
%matplotlib widget
import matplotlib.pyplot as plt
from thesis_code.recording import SpikeRecorder, PerformanceRecorder, GradientRecorder
from thesis_code.plotting import plot_performance, plot_gradients, InteractiveSpikePlot
from thesis_code.runs import load_run
from thesis_code import SRNN
```

```python
cfg, net_state_dict, recorders = load_run(f"{run_dir}/run.pt.gz", {
    "spk_rec": SpikeRecorder,
    "perf_rec": PerformanceRecorder,
    "grad_rec": GradientRecorder,
})
net = SRNN(**cfg["net"])
net.load_state_dict(net_state_dict)
net.eval()
```

```python
perf_rec: PerformanceRecorder = recorders.get("perf_rec")
if perf_rec is not None:
    plot_performance(perf_rec)
    plt.show()
else:
    print("Performance was not recorded for this experiment")
```

```python
grad_rec: GradientRecorder = recorders.get("grad_rec")
if grad_rec is not None:
    plot_gradients(grad_rec)
    plt.show()
else:
    print("Gradients were not recorded for this experiment")
```

```python
spike_rec: SpikeRecorder = recorders.get("spike_rec")
if spike_rec is not None:
    isp = InteractiveSpikePlot(
        recorder=spike_rec, 
        iteration_i_start=0, 
        batch_item_i=0,
    )
    isp.draw()
    plt.show()
else:
    print("Spikes were not recorded for this experiment")
```
