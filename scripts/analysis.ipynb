```python
run_dir = "work"
interactive = True    # the workflow overrides this to False
```

```python
if interactive:
    %matplotlib widget
else:
    %matplotlib inline
import matplotlib.pyplot as plt
from thesis_code.recording import SpikeRecorder, PerformanceRecorder, GradientRecorder
from thesis_code.plotting import plot_performance, plot_gradients, InteractiveSpikePlot
from thesis_code.runs import load_run
from thesis_code import SRNN
```

```python
cfg, net_state_dict, recs = load_run(f"{run_dir}/run.pt.gz")
net = SRNN(**cfg["net"])
net.load_state_dict(net_state_dict)
net.eval()
```

```python
if "perf_rec" in recs.keys():
    perf_rec = PerformanceRecorder.from_state(recs["perf_rec"])
    plot_performance(perf_rec)
    plt.show()
else:
    print("Performance was not recorded for this experiment")
```

```python
if "grad_rec" in recs.keys():
    grad_rec = GradientRecorder.from_state(recs["grad_rec"])
    plot_gradients(grad_rec)
    plt.show()
else:
    print("Gradients were not recorded for this experiment")
```

```python
if interactive:
    if "spike_rec" in recs.keys():
        spike_rec = SpikeRecorder.from_state(recs["spike_rec"])
        isp = InteractiveSpikePlot(
            recorder=spike_rec, 
            iteration_i_start=0, 
            batch_item_i=0,
        )
        isp.draw()
        plt.show()
    else:
        print("Spikes were not recorded for this experiment")
```
